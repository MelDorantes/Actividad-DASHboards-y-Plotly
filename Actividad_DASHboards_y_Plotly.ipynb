####**Descripción del contenido del dataset:**

El dataset describe a clientes de una empresa minorista, incluyendo:

Variables demográficas: edad, estado civil, nivel de educación, ingresos.

Variables del hogar: número de hijos, adolescentes, composición familiar.

Comportamiento de compra: gasto en diferentes categorías (vino, fruta, carne, oro, productos diarios).

Relación con campañas de marketing: si aceptaron o no cada una de las 5 campañas.

Historia con el servicio: quejas, frecuencia de contacto, compras en línea, en tienda, catálogo.

Variable objetivo potencial: Response (si respondió o no a la campaña más reciente).

Es un dataset muy utilizado para análisis exploratorios, dashboards, segmentación de clientes y visualizaciones avanzadas con Plotly.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.figure_factory as ff

# Cargar el archivo
df = pd.read_csv('marketing_campaign.csv')

# Mostrar primeras filas
df.head()
```

##Esquemas de color (Monocromático, Complementario, Triádico)

```python
color_schemes = {
    "monocromatico": ['#264653', '#2A9D8F', '#E9C46A'],
    "complementario": ['#1D3557', '#E63946'],
    "triadico": ['#457B9D', '#E9C46A', '#F4A261']
}

color_schemes
```

##10 gráficos

###Distribución de la Edad

```python
# Crear variable edad
df['Age'] = 2025 - df['Year_Birth']

fig1 = px.histogram(
    df,
    x = 'Age',
    nbins = 40,
    color_discrete_sequence = color_schemes['monocromatico'],
    marginal = 'rug',
    title = 'Distribución de la Edad de los Clientes'
    )
fig1.show()
```

La distribución muestra un rango de edades concentrado entre los 30 y 60 años, lo que indica que la base de clientes se encuentra principalmente en un segmento adulto activo, con potencial adquisitivo estable.

###Ingresos por Nivel Educativo

```python
fig2 = px.box(
    df,
    x = 'Education',
    y = 'Income',
    color = 'Education',
    color_discrete_sequence = color_schemes['complementario'],
    title = 'Distribución de Ingresos por Nivel Educativo'
)
fig2.show()
```

Se observa un incremento progresivo del ingreso conforme aumenta el nivel educativo. Los valores atípicos sugieren grupos con ingresos significativamente superiores dentro de un mismo nivel académico.

###Ingresos vs. Edad

```python
fig3 = px.scatter(
    df,
    x = 'Age',
    y = 'Income',
    color = 'Age',
    color_continuous_scale = color_schemes['triadico'],
    hover_data = ['Marital_Status', 'Education'],
    title = 'Relación entre Edad e Ingresos'
    )
fig3.show()
```

No se aprecia una relación lineal directa: los ingresos tienden a dispersarse a partir de los 35 años. La variabilidad sugiere que factores externos (familia, educación, ocupación) influyen más en el ingreso que la edad en sí misma.

###Compras Totales por Cliente

```python
df['Total_Spending'] = df[[
    'MntWines','MntFruits','MntMeatProducts',
    'MntFishProducts','MntSweetProducts','MntGoldProds'
    ]].sum(axis = 1)

fig4 = px.histogram(
    df,
    x = 'Total_Spending',
    nbins = 50,
    color_discrete_sequence = color_schemes['monocromatico'],
    title = 'Distribución del Gasto Total por Cliente'
    )
fig4.show()
```

La mayoría de los clientes concentra su gasto acumulado en un nivel medio, con un grupo reducido de clientes “premium” que representan compras significativamente mayores.

###Gasto por Categoría

```python
category_means = df[[
    'MntWines','MntFruits','MntMeatProducts',
    'MntFishProducts','MntSweetProducts','MntGoldProds'
    ]].mean().reset_index()

category_means.columns = ['Category', 'Mean_Spend']

fig5 = px.bar(
    category_means,
    x = 'Category',
    y = 'Mean_Spend',
    color = 'Category',
    color_discrete_sequence = color_schemes['complementario'],
    title = 'Gasto Promedio por Categoría'
    )
fig5.show()
```

El vino domina ampliamente el gasto promedio, seguido por productos de carne. Las categorías de fruta, dulces y pescado representan un gasto moderado.
